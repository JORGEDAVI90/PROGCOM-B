```python
import java.util.Scanner;

public class Floristeria {

    static Scanner sc = new Scanner(System.in);

    static String[] flores = {"rosa", "tulipan", "girasol", "amapola", "lirio"};
    static int[] cantidades = {10, 17, 3, 8, 5};
    static int[] precios = {4000, 80000, 5000, 15000, 23000};
    static int[] carrito = {0, 0, 0, 0, 0};

    static double IVA = 0.19;
    static double totalPagar = 0;

    public static void mostrarMenu() {
        System.out.println("\n¿Qué desea hacer?");
        System.out.println("1. Comprar flores");
        System.out.println("2. Pagar");
        System.out.println("3. Generar factura con IVA");
        System.out.println("4. Quitar del carrito");
        System.out.println("5. Salir");
    }

    public static void comprarFlores() {
        System.out.println("\nHa seleccionado: Comprar flores");
        System.out.println("--- STOCK ACTUAL - Precio sin IVA ---");

        for (int i = 0; i < flores.length; i++) {
            if (cantidades[i] > 0) {
                System.out.println(flores[i] + " | Cant: " +
                    cantidades[i] + " | $" + precios[i] + " + IVA");
            }
        }

        System.out.print("\nSeleccione la flor que desea comprar: ");
        String flor = sc.next().toLowerCase();

        int posicion = -1;

        for (int i = 0; i < flores.length; i++) {
            if (flores[i].equals(flor)) {
                posicion = i;
            }
        }

        if (posicion != -1 && cantidades[posicion] > 0) {
            System.out.print("Ingrese la cantidad que desea comprar: ");
            int cantidad = sc.nextInt();

            if (cantidad > 0 && cantidad <= cantidades[posicion]) {
                carrito[posicion] += cantidad;
                cantidades[posicion] -= cantidad;

                System.out.println("¡Se han añadido " + cantidad +
                    " " + flor + "(s) al carrito!");
            } else {
                System.out.println("Cantidad no disponible");
            }
        } else {
            System.out.println("La flor no está disponible");
        }
    }

    public static boolean carritoVacio() {
        for (int i = 0; i < carrito.length; i++) {
            if (carrito[i] > 0) {
                return false;
            }
        }
        return true;
    }

    public static void pagar() {
        if (carritoVacio()) {
            System.out.println("El carrito de compras está vacío.");
            return;
        }

        System.out.println("\n** TABLA DE COMPRAS **");

        double subtotal = 0;

        System.out.println("Flor | Cantidad | Precio | Subtotal");

        for (int i = 0; i < flores.length; i++) {
            if (carrito[i] > 0) {
                int sub = carrito[i] * precios[i];
                subtotal += sub;

                System.out.println(flores[i] + " | " + carrito[i] +
                    " | $" + precios[i] + " | $" + sub);
            }
        }

        double valorIVA = subtotal * IVA;
        totalPagar = subtotal + valorIVA;

        System.out.println("Subtotal sin IVA: $" + subtotal);
        System.out.println("IVA (19%): $" + valorIVA);
        System.out.println("TOTAL A PAGAR: $" + totalPagar);

        System.out.println("\nSeleccione el medio de pago:");
        System.out.println("1. Efectivo");
        System.out.println("2. Tarjeta de crédito");
        System.out.println("3. Tarjeta de débito");

        int opcion = sc.nextInt();

        if (opcion == 1 || opcion == 3) {
            System.out.print("¿Cuánto dinero tiene? ");
            double dinero = sc.nextDouble();

            if (dinero >= totalPagar) {
                System.out.println("Pago exitoso. Cambio: $" +
                    (dinero - totalPagar));
            } else {
                System.out.println("¡No le alcanza! Necesita $" +
                    totalPagar);
            }

        } else if (opcion == 2) {
            System.out.print("¿Número de cuotas? ");
            int cuotas = sc.nextInt();

            if (cuotas > 0) {
                System.out.println("Se cobrará en " + cuotas +
                    " cuotas de $" + (totalPagar / cuotas));
            } else {
                System.out.println("Número de cuotas inválido");
            }
        } else {
            System.out.println("Medio de pago incorrecto");
        }
    }

    public static void generarFactura() {
        System.out.println("\nHa seleccionado: Generar factura con IVA");

        if (carritoVacio()) {
            System.out.println("El carrito de compras está vacío.");
            return;
        }

        double subtotal = 0;

        System.out.println("Flor | Cantidad | Precio | Subtotal");

        for (int i = 0; i < flores.length; i++) {
            if (carrito[i] > 0) {
                int sub = carrito[i] * precios[i];
                subtotal += sub;

                System.out.println(flores[i] + " | " + carrito[i] +
                    " | $" + precios[i] + " | $" + sub);
            }
        }

        double valorIVA = subtotal * IVA;
        double total = subtotal + valorIVA;

        System.out.println("SUBTOTAL: $" + subtotal);
        System.out.println("IVA 19%: $" + valorIVA);
        System.out.println("TOTAL FACTURA: $" + total);
    }

    public static void quitarDelCarrito() {
        System.out.println("\nHa seleccionado: Quitar del carrito");

        if (carritoVacio()) {
            System.out.println("Carrito vacío");
            return;
        }

        for (int i = 0; i < flores.length; i++) {
            if (carrito[i] > 0) {
                System.out.println(flores[i] + ": " + carrito[i]);
            }
        }

        System.out.print("¿Qué flor desea quitar? ");
        String flor = sc.next().toLowerCase();

        int posicion = -1;

        for (int i = 0; i < flores.length; i++) {
            if (flores[i].equals(flor) && carrito[i] > 0) {
                posicion = i;
            }
        }

        if (posicion != -1) {
            System.out.print("Usted compró " + carrito[posicion] +
                ". ¿Cuántas desea quitar? ");
            int cantidad = sc.nextInt();

            if (cantidad > 0 && cantidad <= carrito[posicion]) {
                carrito[posicion] -= cantidad;
                cantidades[posicion] += cantidad;

                System.out.println("Se retiró " + cantidad +
                    " " + flor + "(s)");
            } else {
                System.out.println("Cantidad inválida");
            }
        } else {
            System.out.println("La flor no está en el carrito");
        }
    }

    public static void main(String[] args) {

        System.out.println("Bienvenido a mi floristería - Precios sin IVA");

        int opcion = 0;

        while (opcion != 5) {
            mostrarMenu();

            System.out.print("Seleccione su opción: ");
            opcion = sc.nextInt();

            if (opcion == 1) {
                comprarFlores();
            } else if (opcion == 2) {
                pagar();
            } else if (opcion == 3) {
                generarFactura();
            } else if (opcion == 4) {
                quitarDelCarrito();
            } else if (opcion == 5) {
                System.out.println("Ha seleccionado: Salir");
            } else {
                System.out.println("Número incorrecto");
            }
        }

        System.out.println("Gracias por emplear nuestros servicios. Feliz día");
        sc.close();
    }
}
```
