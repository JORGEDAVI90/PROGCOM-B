```python
nombres = []
fechas = []
horas = []

opcion = 0

while opcion != 9:

    print("1. Ver agenda")
    print("2. Agregar actividad")
    print("3. Buscar actividad")
    print("4. Eliminar actividad")
    print("5. Mostrar cantidad de actividades")
    print("6. Ver primera actividad")
    print("7. Ver ultima actividad")
    print("8. Ordenar agenda")
    print("9. Salir")

    opcion = int(input("Elige una opcion: "))

    if opcion == 1:
        print("--- AGENDA ---")
        i = 0
        while i < len(nombres):
            print(nombres[i], fechas[i], horas[i])
            i = i + 1

    elif opcion == 2:
        nombre = input("Nombre: ")
        fecha = input("Fecha: ")
        hora = input("Hora: ")
        nombres.append(nombre)
        fechas.append(fecha)
        horas.append(hora)
        print("Actividad agregada")

    elif opcion == 3:
        buscar = input("Que actividad buscas: ")
        i = 0
        while i < len(nombres):
            if nombres[i] == buscar:
                print(nombres[i], fechas[i], horas[i])
            i = i + 1

    elif opcion == 4:
        borrar = input("Que actividad quieres eliminar: ")
        i = 0
        while i < len(nombres):
            if nombres[i] == borrar:
                nombres.pop(i)
                fechas.pop(i)
                horas.pop(i)
                print("Actividad eliminada")
                break
            i = i + 1

    elif opcion == 5:
        print("Cantidad de actividades:", len(nombres))

    elif opcion == 6:
        if len(nombres) > 0:
            print(nombres[0], fechas[0], horas[0])
        else:
            print("No hay actividades")

    elif opcion == 7:
        if len(nombres) > 0:
            print(nombres[-1], fechas[-1], horas[-1])
        else:
            print("No hay actividades")

    elif opcion == 8:
        nombres.sort()
        print("Agenda ordenada")

    elif opcion == 9:
        print("Adios")
```
