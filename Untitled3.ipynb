```python
import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

OPCIONES = ["piedra", "papel", "tijera", "lagarto", "spock"]

EMOJI = {
    "piedra": "✊",
    "papel": "✋",
    "tijera": "✌️",
    "lagarto": "🦎",
    "spock": "🖖",
}

REGLAS = {
    "tijera":  {"papel": "corta", "lagarto": "decapita"},
    "papel":   {"piedra": "envuelve", "spock": "desautoriza"},
    "piedra":  {"lagarto": "aplasta", "tijera": "aplasta"},
    "lagarto": {"spock": "envenena", "papel": "devora"},
    "spock":   {"tijera": "rompe", "piedra": "vaporiza"},
}

puntos = {"jugador": 0, "maquina": 0}


def determinar_ganador(jugador, maquina):
    if jugador == maquina:
        return "empate"
    if maquina in REGLAS[jugador]:
        return "jugador"
    return "maquina"


def frase_resultado(jugador, maquina, resultado):
    if resultado == "empate":
        return f"Empate: ambos eligieron {jugador.capitalize()}."
    if resultado == "jugador":
        verbo = REGLAS[jugador][maquina]
        return f"{jugador.capitalize()} {verbo} a {maquina.capitalize()}. ¡Ganaste!"
    verbo = REGLAS[maquina][jugador]
    return f"{maquina.capitalize()} {verbo} a {jugador.capitalize()}. Ganó la máquina."


# --- Widgets ---
titulo = widgets.HTML("<h2>Piedra · Papel · Tijera · Lagarto · Spock</h2>")
marcador = widgets.HTML(f"<b>Tú: {puntos['jugador']}   Máquina: {puntos['maquina']}</b>")
resultado_lbl = widgets.HTML("Elige una opción para empezar.")

botones = [
    widgets.Button(description=f"{EMOJI[o]} {o.capitalize()}", layout=widgets.Layout(width="120px", height="50px"))
    for o in OPCIONES
]

reiniciar_btn = widgets.Button(description="Reiniciar marcador", button_style="danger")


def jugar(boton):
    opcion_jugador = OPCIONES[botones.index(boton)]
    opcion_maquina = random.choice(OPCIONES)
    resultado = determinar_ganador(opcion_jugador, opcion_maquina)

    if resultado == "jugador":
        puntos["jugador"] += 1
    elif resultado == "maquina":
        puntos["maquina"] += 1

    texto = (
        f"Tú: {EMOJI[opcion_jugador]} {opcion_jugador.capitalize()}"
        f"   vs   Máquina: {EMOJI[opcion_maquina]} {opcion_maquina.capitalize()}<br>"
        f"{frase_resultado(opcion_jugador, opcion_maquina, resultado)}"
    )
    resultado_lbl.value = texto
    marcador.value = f"<b>Tú: {puntos['jugador']}   Máquina: {puntos['maquina']}</b>"


def reiniciar(boton):
    puntos["jugador"] = 0
    puntos["maquina"] = 0
    marcador.value = f"<b>Tú: 0   Máquina: 0</b>"
    resultado_lbl.value = "Marcador reiniciado. ¡Elige una opción!"


for b in botones:
    b.on_click(jugar)
reiniciar_btn.on_click(reiniciar)

display(titulo, marcador, widgets.HBox(botones), resultado_lbl, reiniciar_btn)
```
