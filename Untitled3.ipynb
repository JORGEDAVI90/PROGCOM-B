```python
import random
from collections import Counter
import ipywidgets as widgets
from IPython.display import display, HTML



EMOJI = {
    "piedra": "✊", "papel": "✋", "tijera": "✌️",
    "fuego": "🔥", "agua": "💧",
    "lagarto": "🦎", "spock": "🖖", "heavy": "🤵", "papa": "⛪",
}


REGLAS_5 = {
    "tijera":  {"papel": "corta", "fuego": "sofoca"},
    "papel":   {"piedra": "envuelve", "agua": "absorbe"},
    "piedra":  {"fuego": "apaga", "tijera": "aplasta"},
    "fuego":   {"agua": "evapora", "papel": "quema"},
    "agua":    {"tijera": "oxida", "piedra": "erosiona"},
}

REGLAS_7 = {
    "piedra":  {"tijera": "aplasta", "lagarto": "machaca", "heavy": "aplasta"},
    "tijera":  {"papel": "corta", "lagarto": "decapita", "heavy": "corta"},
    "papel":   {"piedra": "envuelve", "spock": "desautoriza", "heavy": "envuelve"},
    "lagarto": {"spock": "envenena", "papel": "devora", "papa": "ignora"},
    "spock":   {"tijera": "rompe", "piedra": "vaporiza", "papa": "calcula y vence a"},
    "papa":    {"piedra": "conoce el truco y vence a", "papel": "conoce el truco y vence a",
                "tijera": "conoce el truco y vence a"},
    "heavy":   {"papa": "moderniza", "lagarto": "aplasta", "spock": "rompe"},
}

VARIANTES = {
    "5 elementos (Piedra·Papel·Tijera·Fuego·Agua)": REGLAS_5,
    "7 elementos (Piedra·Papel·Tijera·Lagarto·Spock·Heavy·Papa)": REGLAS_7,
}



def construir_contra(reglas):
    """A partir de REGLAS (quién le gana a quién) arma CONTRA (quién le pierde a quién)."""
    contra = {o: [] for o in reglas}
    for atacante, victimas in reglas.items():
        for victima in victimas:
            contra.setdefault(victima, []).append(atacante)
    return contra


def validar_generalizacion(reglas):
    """Verifica las condiciones del paper de Jordán & Sanabria-Codesal (2013)."""
    opciones = list(reglas.keys())
    n = len(opciones)
    contra = construir_contra(reglas)
    esperado = (n - 1) // 2

    mensajes = [f"n = {n} elementos → ¿impar? {'✅ sí' if n % 2 == 1 else '❌ no'}"]
    balanceado = True
    for o in opciones:
        wins = len(reglas.get(o, {}))
        losses = len(contra.get(o, []))
        ok = (wins == esperado and losses == esperado)
        balanceado &= ok
        mensajes.append(
            f"  {o.capitalize():8s}: gana a {wins} y pierde con {losses} "
            f"(esperado {esperado} y {esperado}) {'✅' if ok else '❌'}"
        )

    valido = (n % 2 == 1) and balanceado
    p_empate = 1 / n
    p_gano_perdio = (n - 1) / (2 * n)
    mensajes.append(
        f"\nSegún la teoría de grafos del paper: grafo dirigido con grado de entrada = "
        f"grado de salida en cada vértice ⇒ {'sin' if valido else 'CON'} estrategia dominante."
    )
    mensajes.append(
        f"Probabilidad teórica → Empate: {p_empate:.1%}   |   Ganar: {p_gano_perdio:.1%}   |   Perder: {p_gano_perdio:.1%}"
    )
    return valido, "\n".join(mensajes)


DIFICULTADES = {
    "Fácil":     {"memoria": 6, "prob_aprender": 0.20, "prob_trampa": 0.0},
    "Normal":    {"memoria": 6, "prob_aprender": 0.55, "prob_trampa": 0.0},
    "Difícil":   {"memoria": 4, "prob_aprender": 0.80, "prob_trampa": 0.10},
    "Imposible": {"memoria": 3, "prob_aprender": 0.95, "prob_trampa": 0.35},
}

estado = {
    "puntos_jugador": 0, "puntos_maquina": 0,
    "racha": 0, "mejor_racha": 0,
    "historial_jugador": [], "log": [],
    "dificultad": "Normal",
    "variante": list(VARIANTES.keys())[0],
}


def reglas_actuales():
    return VARIANTES[estado["variante"]]


def contra_actual():
    return construir_contra(reglas_actuales())


def elegir_maquina(ultima_jugador=None):
    cfg = DIFICULTADES[estado["dificultad"]]
    contra = contra_actual()
    opciones = list(reglas_actuales().keys())
    historial = estado["historial_jugador"]

    if ultima_jugador and random.random() < cfg["prob_trampa"]:
        return random.choice(contra[ultima_jugador])
    if len(historial) >= 3 and random.random() < cfg["prob_aprender"]:
        favorita, _ = Counter(historial[-cfg["memoria"]:]).most_common(1)[0]
        return random.choice(contra[favorita])
    return random.choice(opciones)


def determinar_ganador(jugador, maquina):
    if jugador == maquina:
        return "empate"
    return "jugador" if maquina in reglas_actuales()[jugador] else "maquina"


def frase_resultado(jugador, maquina, resultado):
    if resultado == "empate":
        return f"⚖️ Empate: ambos eligieron {jugador.capitalize()}."
    if resultado == "jugador":
        verbo = reglas_actuales()[jugador][maquina]
        return f"🏆 {jugador.capitalize()} {verbo} a {maquina.capitalize()}. ¡Ganaste!"
    verbo = reglas_actuales()[maquina][jugador]
    return f"💀 {maquina.capitalize()} {verbo} a {jugador.capitalize()}. Ganó la máquina."



titulo = widgets.HTML("<h2>Piedra, Papel, Tijera... y sus generalizaciones 📐</h2>"
                       "<p style='color:gray'>Basado en Jordán &amp; Sanabria-Codesal (2013), MSEL 6(2)</p>")

variante_selector = widgets.Dropdown(options=list(VARIANTES.keys()), description="Variante:",
                                      style={"description_width": "initial"}, layout=widgets.Layout(width="480px"))
dificultad_selector = widgets.ToggleButtons(options=list(DIFICULTADES.keys()), value="Normal", description="Dificultad:")
math_btn = widgets.Button(description="📐 Ver validación matemática", button_style="info")

marcador = widgets.HTML()
racha_lbl = widgets.HTML()
resultado_lbl = widgets.HTML("Elige variante y dificultad, luego juega.")
historial_lbl = widgets.HTML()
math_output = widgets.HTML()
botones_caja = widgets.HBox([])
reiniciar_btn = widgets.Button(description="🔄 Reiniciar todo", button_style="danger")


def construir_botones():
    opciones = list(reglas_actuales().keys())
    nuevos = [
        widgets.Button(description=f"{EMOJI[o]} {o.capitalize()}", layout=widgets.Layout(width="110px", height="50px"))
        for o in opciones
    ]
    for b in nuevos:
        b.on_click(jugar)
    botones_caja.children = nuevos


def actualizar_marcador():
    n = len(reglas_actuales())
    marcador.value = (
        f"<b>Tú: {estado['puntos_jugador']}   |   Máquina: {estado['puntos_maquina']}   "
        f"|   Variante: {n} elementos   |   Dificultad: {estado['dificultad']}</b>"
    )
    fuego_racha = "🔥" * min(estado["racha"], 5)
    racha_lbl.value = f"Racha actual: {estado['racha']} {fuego_racha}   (mejor racha: {estado['mejor_racha']})"
    ultimas = estado["log"][-5:]
    historial_lbl.value = "Últimas jugadas: " + " ".join(
        f"[{EMOJI[j]}vs{EMOJI[m]}:{'✅' if r=='jugador' else '❌' if r=='maquina' else '➖'}]"
        for j, m, r in ultimas
    ) if ultimas else ""


def jugar(boton):
    opcion_jugador = [k for k in reglas_actuales() if EMOJI[k] in boton.description][0]
    # más robusto: mapear por descripción
    opcion_jugador = boton.description.split(" ", 1)[1].lower()

    ultima = estado["historial_jugador"][-1] if estado["historial_jugador"] else None
    opcion_maquina = elegir_maquina(ultima)
    resultado = determinar_ganador(opcion_jugador, opcion_maquina)

    estado["historial_jugador"].append(opcion_jugador)
    estado["log"].append((opcion_jugador, opcion_maquina, resultado))

    bonus_texto = ""
    if resultado == "jugador":
        estado["puntos_jugador"] += 1
        estado["racha"] += 1
        if estado["racha"] % 3 == 0:
            estado["puntos_jugador"] += 1
            bonus_texto = " <b>🎉 ¡Combo x3! +1 punto extra</b>"
        estado["mejor_racha"] = max(estado["mejor_racha"], estado["racha"])
    elif resultado == "maquina":
        estado["puntos_maquina"] += 1
        estado["racha"] = 0

    texto = (
        f"Tú: {EMOJI[opcion_jugador]} {opcion_jugador.capitalize()}"
        f"   vs   Máquina: {EMOJI[opcion_maquina]} {opcion_maquina.capitalize()}<br>"
        f"{frase_resultado(opcion_jugador, opcion_maquina, resultado)}{bonus_texto}"
    )
    resultado_lbl.value = texto
    actualizar_marcador()


def cambiar_variante(change):
    estado["variante"] = change["new"]
    estado["historial_jugador"] = []
    estado["log"] = []
    construir_botones()
    actualizar_marcador()
    math_output.value = ""
    resultado_lbl.value = f"Cambiaste a la variante de {len(reglas_actuales())} elementos. ¡Juega!"


def cambiar_dificultad(change):
    estado["dificultad"] = change["new"]
    actualizar_marcador()


def mostrar_validacion(boton):
    valido, texto = validar_generalizacion(reglas_actuales())
    color = "green" if valido else "red"
    texto_html = texto.replace("\n", "<br>")
    math_output.value = f"<pre style='background:#f5f5f5;padding:10px;border-left:4px solid {color};'>{texto_html}</pre>"


def reiniciar(boton):
    estado.update({"puntos_jugador": 0, "puntos_maquina": 0, "racha": 0, "mejor_racha": 0,
                    "historial_jugador": [], "log": []})
    resultado_lbl.value = "Marcador reiniciado. ¡Elige una opción!"
    actualizar_marcador()


variante_selector.observe(cambiar_variante, names="value")
dificultad_selector.observe(cambiar_dificultad, names="value")
math_btn.on_click(mostrar_validacion)
reiniciar_btn.on_click(reiniciar)

construir_botones()
actualizar_marcador()
display(titulo, variante_selector, dificultad_selector, marcador, racha_lbl,
        botones_caja, resultado_lbl, historial_lbl, widgets.HBox([math_btn, reiniciar_btn]), math_output)
```
