```python
def dinero(valor):
    return f"${valor:,.2f}"

def calcular_prestamo():
    print("\n" + "=" * 55)
    print("              SIMULADOR DE PRÉSTAMOS")
    print("=" * 55)
    try:
        monto = float(input("Ingrese el valor del préstamo: $"))
        tasa_ea = float(input("Ingrese la tasa efectiva anual (% E.A.): "))
        meses = int(input("Ingrese el plazo del préstamo en meses: "))
        if monto <= 0 or tasa_ea < 0 or meses <= 0:
            print("\nError: los valores ingresados no son válidos.")
            return

        tasa_anual = tasa_ea / 100
        tasa_mensual = (1 + tasa_anual) ** (1 / 12) - 1

        if tasa_mensual == 0:
            cuota = monto / meses
        else:
            cuota = (
                monto
                * tasa_mensual
                * (1 + tasa_mensual) ** meses
                / ((1 + tasa_mensual) ** meses - 1)
            )
        total_pagado = cuota * meses
        intereses = total_pagado - monto
        print("\n------------- RESULTADOS -------------")
        print("Valor del préstamo:", dinero(monto))
        print(f"Tasa E.A.: {tasa_ea:.2f}%")
        print(f"Tasa efectiva mensual: {tasa_mensual * 100:.4f}%")
        print(f"Plazo: {meses} meses")
        print("Cuota mensual:", dinero(cuota))
        print("Total pagado:", dinero(total_pagado))
        print("Total de intereses:", dinero(intereses))
        opcion = input(
            "\n¿Desea ver la tabla de amortización? (S/N): "
        ).upper()
        if opcion == "S":
            tabla_amortizacion(monto, tasa_mensual, meses, cuota)
    except ValueError:
        print("\nError: debe ingresar únicamente valores numéricos.")

def tabla_amortizacion(monto, tasa, meses, cuota):
    saldo = monto
    print("\n" + "=" * 92)
    print("                        TABLA DE AMORTIZACIÓN")
    print("=" * 92)
    print(
        f"{'Mes':<6}"
        f"{'Saldo inicial':>18}"
        f"{'Interés':>16}"
        f"{'Abono capital':>18}"
        f"{'Cuota':>16}"
        f"{'Saldo final':>18}"
    )
    print("-" * 92)
    for mes in range(1, meses + 1):
        saldo_inicial = saldo
        interes = saldo_inicial * tasa
        abono_capital = cuota - interes
        saldo_final = saldo_inicial - abono_capital

        if saldo_final < 0:
            saldo_final = 0
        print(
            f"{mes:<6}"
            f"{dinero(saldo_inicial):>18}"
            f"{dinero(interes):>16}"
            f"{dinero(abono_capital):>18}"
            f"{dinero(cuota):>16}"
            f"{dinero(saldo_final):>18}"
        )
        saldo = saldo_final

def interes_compuesto():
    print("\n" + "=" * 55)
    print("          CALCULADORA DE INTERÉS COMPUESTO")
    print("=" * 55)
    try:
        capital = float(input("Ingrese el capital inicial: $"))
        tasa = float(input("Ingrese la tasa efectiva anual (%): "))
        anos = float(input("Ingrese el tiempo de inversión en años: "))
        if capital <= 0 or tasa < 0 or anos < 0:
            print("\nError: los valores ingresados no son válidos.")
            return
        tasa_decimal = tasa / 100
        valor_futuro = capital * (1 + tasa_decimal) ** anos
        ganancia = valor_futuro - capital
        print("\n------------- RESULTADOS -------------")
        print("Capital inicial:", dinero(capital))
        print(f"Tasa anual: {tasa:.2f}%")
        print(f"Tiempo: {anos:g} años")
        print("Valor futuro:", dinero(valor_futuro))
        print("Ganancia obtenida:", dinero(ganancia))
    except ValueError:
        print("\nError: debe ingresar únicamente valores numéricos.")

def calcular_rentabilidad():
    print("\n" + "=" * 55)
    print("          RENTABILIDAD DE UNA INVERSIÓN")
    print("=" * 55)
    try:
        inversion_inicial = float(
            input("Ingrese el valor inicial de la inversión: $")
        )
        valor_final = float(
            input("Ingrese el valor final de la inversión: $")
        )
        if inversion_inicial <= 0 or valor_final < 0:
            print("\nError: los valores ingresados no son válidos.")
            return
        utilidad = valor_final - inversion_inicial
        rentabilidad = (
            utilidad / inversion_inicial
        ) * 100
        print("\n------------- RESULTADOS -------------")
        print("Inversión inicial:", dinero(inversion_inicial))
        print("Valor final:", dinero(valor_final))
        print("Ganancia/Pérdida:", dinero(utilidad))
        print(f"Rentabilidad: {rentabilidad:.2f}%")
        if rentabilidad > 0:
            print("Resultado: la inversión generó una ganancia.")
        elif rentabilidad == 0:
            print("Resultado: la inversión no generó ganancia ni pérdida.")
        else:
            print("Resultado: la inversión generó una pérdida.")
    except ValueError:
        print("\nError: debe ingresar únicamente valores numéricos.")

def comparar_inversiones():
    print("\n" + "=" * 55)
    print("             COMPARADOR DE INVERSIONES")
    print("=" * 55)
    try:
        capital = float(
            input("Ingrese el capital disponible para invertir: $")
        )
        print("\n--- INVERSIÓN A ---")
        tasa_a = float(
            input("Ingrese la tasa efectiva anual de A (%): ")
        )
        tiempo_a = float(
            input("Ingrese el tiempo de la inversión A en años: ")
        )
        print("\n--- INVERSIÓN B ---")
        tasa_b = float(
            input("Ingrese la tasa efectiva anual de B (%): ")
        )
        tiempo_b = float(
            input("Ingrese el tiempo de la inversión B en años: ")
        )
        if (
            capital <= 0
            or tasa_a < 0
            or tasa_b < 0
            or tiempo_a < 0
            or tiempo_b < 0
        ):
            print("\nError: los valores ingresados no son válidos.")
            return
        valor_a = capital * (1 + tasa_a / 100) ** tiempo_a
        valor_b = capital * (1 + tasa_b / 100) ** tiempo_b
        ganancia_a = valor_a - capital
        ganancia_b = valor_b - capital
        print("\n" + "=" * 55)
        print("                  RESULTADOS")
        print("=" * 55)
        print("\nINVERSIÓN A")
        print(f"Tasa: {tasa_a:.2f}% E.A.")
        print(f"Tiempo: {tiempo_a:g} años")
        print("Valor futuro:", dinero(valor_a))
        print("Ganancia:", dinero(ganancia_a))
        print("\nINVERSIÓN B")
        print(f"Tasa: {tasa_b:.2f}% E.A.")
        print(f"Tiempo: {tiempo_b:g} años")
        print("Valor futuro:", dinero(valor_b))
        print("Ganancia:", dinero(ganancia_b))
        print("\n------------- COMPARACIÓN -------------")
        if valor_a > valor_b:
            diferencia = valor_a - valor_b
            print("La inversión A genera un mayor valor futuro.")
            print(
                "Diferencia respecto a B:",
                dinero(diferencia)
            )
        elif valor_b > valor_a:
            diferencia = valor_b - valor_a
            print("La inversión B genera un mayor valor futuro.")
            print(
                "Diferencia respecto a A:",
                dinero(diferencia)
            )
        else:
            print(
                "Las dos inversiones generan el mismo valor futuro."
            )
    except ValueError:
        print(
            "\nError: debe ingresar únicamente valores numéricos."
        )

def calcular_valor_presente():
    print("\n" + "=" * 55)
    print("              CALCULADORA DE VALOR PRESENTE")
    print("=" * 55)
    try:
        valor_futuro = float(
            input("Ingrese el dinero que recibirá en el futuro: $")
        )
        tasa = float(
            input("Ingrese la tasa de descuento anual (%): ")
        )
        tiempo = float(
            input("Ingrese el número de años: ")
        )
        if valor_futuro <= 0 or tasa < 0 or tiempo < 0:
            print("\nError: los valores ingresados no son válidos.")
            return
        tasa_decimal = tasa / 100
        valor_presente = (
            valor_futuro
            / ((1 + tasa_decimal) ** tiempo)
        )
        descuento = valor_futuro - valor_presente
        print("\n------------- RESULTADOS -------------")
        print("Valor futuro:", dinero(valor_futuro))
        print(f"Tasa de descuento: {tasa:.2f}%")
        print(f"Tiempo: {tiempo:g} años")
        print(
            "Valor presente:",
            dinero(valor_presente)
        )
        print(
            "Diferencia entre valor futuro y presente:",
            dinero(descuento)
        )
    except ValueError:
        print(
            "\nError: debe ingresar únicamente valores numéricos."
        )


def menu():
    while True:
        print("\n")
        print("=" * 60)
        print("                       FINANPY")
        print("              SISTEMA DE ANÁLISIS FINANCIERO")
        print("=" * 60)
        print("\n1. Simulador de préstamos")
        print("2. Calcular interés compuesto")
        print("3. Calcular rentabilidad de una inversión")
        print("4. Comparar dos inversiones")
        print("5. Calcular valor presente")
        print("6. Salir")
        print("-" * 60)
        opcion = input(
            "Seleccione una opción (1-6): "
        )
        if opcion == "1":
            calcular_prestamo()
        elif opcion == "2":
            interes_compuesto()
        elif opcion == "3":
            calcular_rentabilidad()
        elif opcion == "4":
            comparar_inversiones()
        elif opcion == "5":
            calcular_valor_presente()
        elif opcion == "6":
            print("\n" + "=" * 55)
            print("Gracias por utilizar FINANPY.")
            print("Programa finalizado correctamente.")
            print("=" * 55)
            break
        else:
            print(
                "\nOpción incorrecta. "
                "Seleccione un número entre 1 y 6."
            )

menu()
```
