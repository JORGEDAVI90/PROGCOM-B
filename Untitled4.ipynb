#Piedra papel o tijera

```python
import random
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets

elementos = [
    "Piedra",
    "Pistola",
    "Trueno",
    "Diablo",
    "Dragón",
    "Agua",
    "Aire",
    "Papel",
    "Esponja",
    "Lobo",
    "Árbol",
    "Humano",
    "Serpiente",
    "Tijera",
    "Fuego"
]

emojis = {
    "Piedra": "🪨",
    "Papel": "📄",
    "Tijera": "✂️",
    "Pistola": "🔫",
    "Trueno": "⚡",
    "Diablo": "😈",
    "Dragón": "🐉",
    "Agua": "💧",
    "Aire": "💨",
    "Esponja": "🧽",
    "Lobo": "🐺",
    "Árbol": "🌳",
    "Humano": "🧑",
    "Serpiente": "🐍",
    "Fuego": "🔥"
}

reglas = {}

for i, elemento in enumerate(elementos):
    reglas[elemento] = []
    for j in range(1, 8):
        reglas[elemento].append(elementos[(i + j) % len(elementos)])

acciones_especiales = {
    ("Piedra", "Tijera"): "La piedra aplasta las tijeras",
    ("Piedra", "Fuego"): "La piedra vence al fuego",
    ("Tijera", "Papel"): "Las tijeras cortan el papel",
    ("Fuego", "Papel"): "El fuego quema el papel",
    ("Fuego", "Esponja"): "El fuego quema la esponja",
    ("Fuego", "Humano"): "El fuego vence al humano",
    ("Agua", "Fuego"): "El agua apaga el fuego",
    ("Agua", "Piedra"): "El agua erosiona la piedra",
    ("Agua", "Tijera"): "El agua oxida las tijeras",
    ("Papel", "Piedra"): "El papel envuelve la piedra",
    ("Esponja", "Agua"): "La esponja absorbe el agua",
    ("Humano", "Agua"): "El humano contamina el agua",
    ("Pistola", "Humano"): "La pistola vence al humano",
    ("Pistola", "Piedra"): "La pistola destruye la piedra"
}

def obtener_accion(ganador, perdedor):
    if (ganador, perdedor) in acciones_especiales:
        return acciones_especiales[(ganador, perdedor)]
    return f"{ganador} vence a {perdedor}"

puntos_jugador = 0
puntos_computadora = 0
empates = 0
rondas = 0
dificultad = "Fácil"

salida = widgets.Output()

selector_dificultad = widgets.ToggleButtons(
    options=[
        ("🟢 Fácil", "Fácil"),
        ("🟡 Medio", "Medio"),
        ("🔴 Difícil", "Difícil")
    ],
    value="Fácil",
    description="Dificultad:",
    layout=widgets.Layout(width="520px")
)

def cambiar_dificultad(cambio):
    global dificultad
    if cambio["name"] == "value":
        dificultad = cambio["new"]

selector_dificultad.observe(cambiar_dificultad, names="value")

def opciones_que_vencen(eleccion):
    return [
        elemento for elemento in elementos
        if eleccion in reglas[elemento]
    ]

def opciones_que_pierden(eleccion):
    return reglas[eleccion]

def eleccion_computadora_segun_dificultad(eleccion_jugador):
    ganadoras = opciones_que_vencen(eleccion_jugador)
    perdedoras = opciones_que_pierden(eleccion_jugador)

    numero = random.random()

    if dificultad == "Fácil":
        return random.choice(elementos)

    elif dificultad == "Medio":
        if numero < 0.50:
            return random.choice(ganadoras)
        elif numero < 0.85:
            return random.choice(perdedoras)
        else:
            return eleccion_jugador

    elif dificultad == "Difícil":
        if numero < 0.75:
            return random.choice(ganadoras)
        elif numero < 0.90:
            return random.choice(perdedoras)
        else:
            return eleccion_jugador

def jugar(eleccion_jugador):
    global puntos_jugador
    global puntos_computadora
    global empates
    global rondas

    eleccion_computadora = eleccion_computadora_segun_dificultad(
        eleccion_jugador
    )

    rondas += 1

    if eleccion_jugador == eleccion_computadora:
        resultado = "¡EMPATE!"
        emoji_resultado = "🤝😮"
        empates += 1
        explicacion = f"Ambos eligieron {eleccion_jugador}."
        color = "#f59e0b"

    elif eleccion_computadora in reglas[eleccion_jugador]:
        resultado = "¡GANASTE!"
        emoji_resultado = "🥳🏆🎉"
        puntos_jugador += 1
        explicacion = obtener_accion(
            eleccion_jugador,
            eleccion_computadora
        )
        color = "#16a34a"

    else:
        resultado = "¡GANÓ LA COMPUTADORA!"
        emoji_resultado = "🤖🏆😎"
        puntos_computadora += 1
        explicacion = obtener_accion(
            eleccion_computadora,
            eleccion_jugador
        )
        color = "#dc2626"

    with salida:
        clear_output(wait=True)

        display(HTML(f"""
        <div style="
            font-family:Arial;
            text-align:center;
            max-width:850px;
            margin:20px auto;
            padding:25px;
            border-radius:22px;
            background:#f8fafc;
            box-shadow:0 4px 16px rgba(0,0,0,0.15);
        ">

            <h2 style="color:#312e81;">
                ⚔️ RONDA {rondas}
            </h2>

            <div style="
                display:inline-block;
                padding:8px 18px;
                background:#ede9fe;
                border-radius:20px;
                color:#5b21b6;
                font-weight:bold;
                margin-bottom:15px;
            ">
                Nivel: {dificultad}
            </div>

            <div style="
                display:flex;
                justify-content:space-around;
                align-items:center;
                margin:25px 0;
            ">

                <div style="
                    background:white;
                    padding:20px;
                    border-radius:18px;
                    min-width:220px;
                ">
                    <h3>👩‍💻 JUGADOR</h3>

                    <div style="font-size:70px;">
                        {emojis[eleccion_jugador]}
                    </div>

                    <h2>{eleccion_jugador}</h2>
                </div>

                <div style="
                    font-size:35px;
                    font-weight:bold;
                    color:#7c3aed;
                ">
                    VS
                </div>

                <div style="
                    background:white;
                    padding:20px;
                    border-radius:18px;
                    min-width:220px;
                ">
                    <h3>🤖 COMPUTADORA</h3>

                    <div style="font-size:70px;">
                        {emojis[eleccion_computadora]}
                    </div>

                    <h2>{eleccion_computadora}</h2>
                </div>

            </div>

            <h1 style="
                color:{color};
                margin-top:20px;
            ">
                {emoji_resultado}
                <br>
                {resultado}
            </h1>

            <p style="
                font-size:20px;
                color:#475569;
            ">
                <b>{explicacion}</b>
            </p>

            <div style="
                background:white;
                padding:18px;
                border-radius:15px;
                margin-top:20px;
            ">

                <h3>📊 MARCADOR</h3>

                <span style="
                    font-size:20px;
                    margin:15px;
                ">
                    👩‍💻 Jugador:
                    <b>{puntos_jugador}</b>
                </span>

                <span style="
                    font-size:20px;
                    margin:15px;
                ">
                    🤖 Computadora:
                    <b>{puntos_computadora}</b>
                </span>

                <span style="
                    font-size:20px;
                    margin:15px;
                ">
                    🤝 Empates:
                    <b>{empates}</b>
                </span>

            </div>

        </div>
        """))

def boton_presionado(boton):
    jugar(boton.elemento)

def reiniciar(boton):
    global puntos_jugador
    global puntos_computadora
    global empates
    global rondas

    puntos_jugador = 0
    puntos_computadora = 0
    empates = 0
    rondas = 0

    with salida:
        clear_output()

        display(HTML(f"""
        <div style="
            text-align:center;
            padding:30px;
            font-family:Arial;
        ">

            <div style="font-size:65px;">
                🎮✨
            </div>

            <h2 style="color:#312e81;">
                ¡Nueva partida!
            </h2>

            <p style="font-size:18px;">
                Nivel seleccionado:
                <b>{dificultad}</b>
            </p>

            <p>
                Elige uno de los 15 elementos para comenzar.
            </p>

        </div>
        """))

botones = []

for elemento in elementos:
    boton = widgets.Button(
        description=f"{emojis[elemento]} {elemento}",
        layout=widgets.Layout(
            width="155px",
            height="52px",
            margin="5px"
        )
    )

    boton.elemento = elemento
    boton.on_click(boton_presionado)
    botones.append(boton)

boton_reiniciar = widgets.Button(
    description="🔄 Reiniciar partida",
    button_style="danger",
    layout=widgets.Layout(
        width="230px",
        height="50px",
        margin="15px"
    )
)

boton_reiniciar.on_click(reiniciar)

display(HTML("""
<div style="
    text-align:center;
    font-family:Arial;
    max-width:950px;
    margin:auto;
">

    <div style="
        background:linear-gradient(135deg,#312e81,#7c3aed);
        color:white;
        padding:25px;
        border-radius:20px;
        margin-bottom:20px;
    ">

        <h1>
            🎮 PIEDRA, PAPEL, TIJERA
        </h1>

        <h2>
            ⚔️ Versión de 15 elementos ⚔️
        </h2>

        <p style="font-size:17px;">
            Elige tu dificultad y enfréntate a la computadora.
        </p>

    </div>

    <div style="
        background:#eef2ff;
        padding:16px;
        border-radius:15px;
        margin-bottom:18px;
    ">

        <h3>🎯 NIVELES DE DIFICULTAD</h3>

        <p>
            🟢 <b>Fácil:</b> la computadora juega al azar.
            <br>
            🟡 <b>Medio:</b> la computadora juega con mayor estrategia.
            <br>
            🔴 <b>Difícil:</b> la computadora tiene mayor posibilidad de elegir una opción ganadora.
        </p>

    </div>

</div>
"""))

display(
    widgets.HBox(
        [selector_dificultad],
        layout=widgets.Layout(
            justify_content="center",
            margin="10px 0 20px 0"
        )
    )
)

filas = []

for i in range(0, len(botones), 5):
    fila = widgets.HBox(
        botones[i:i+5],
        layout=widgets.Layout(
            justify_content="center"
        )
    )
    filas.append(fila)

interfaz_botones = widgets.VBox(
    filas,
    layout=widgets.Layout(
        align_items="center"
    )
)

display(interfaz_botones)

display(
    widgets.HBox(
        [boton_reiniciar],
        layout=widgets.Layout(
            justify_content="center"
        )
    )
)

display(salida)

with salida:
    display(HTML("""
    <div style="
        text-align:center;
        padding:25px;
        font-family:Arial;
        color:#475569;
    ">

        <div style="font-size:60px;">
            🎯
        </div>

        <h2>
            ¡Selecciona la dificultad y elige tu elemento!
        </h2>

        <p style="font-size:17px;">
            Tienes 15 opciones disponibles.
            <br>
            🪨 📄 ✂️ 🔫 ⚡ 😈 🐉 💧 💨 🧽 🐺 🌳 🧑 🐍 🔥
        </p>

    </div>
    """))
```
