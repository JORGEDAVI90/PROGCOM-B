```python
flores={"tipo":["rosa","tulipan","girasol","amapola","lirio"],
        "cantidad": [1,17,3,8,5],
        "precio": [4000,8000,5000,15000,23000]
        }

print("Bienvenido a mi floristería\n")
print("menu")

carrito = []
total = 0

while True:
  print("Qué desea hacer?")
  print("1. Visualizar nuestras flores")
  print("2. comprar")
  print("3. pagar")
  print("4. factura de compra")
  print("5. salir")

  opc=int(input("ingrese su opción\n"))
  match opc:
    case 1:
      print("estas son nuestra flores")
      print(f'{"Flores":<15} | {"Cantidad":<10} | {"Precio unidad":<15}')
      for tipo,cantidad,precio in zip(flores["tipo"],flores["cantidad"],flores["precio"]):
        print(f"{tipo.capitalize():<15} | {cantidad:<10} | {precio:<15}")

      pass

    case 2:
      flor = input("Ingrese el nombre de la flor que desea comprar\n").lower()

      if flor in flores["tipo"]:
        pos = flores["tipo"].index(flor)

        cantidad = int(input("Ingrese la cantidad que desea comprar\n"))

        if cantidad <= flores["cantidad"][pos]:
          precio = flores["precio"][pos]
          subtotal = cantidad * precio

          carrito.append([flor, cantidad, precio, subtotal])

          flores["cantidad"][pos] = flores["cantidad"][pos] - cantidad

          total = total + subtotal

          print("Compra agregada al carrito")
          print("Subtotal: $", subtotal)

        else:
          print("No hay suficiente cantidad de esta flor")

      else:
        print("La flor no existe")

      pass

    case 3:

      carrito={"tipo":["girasol","lirio"],
               "cantidad":[5,13],
               "precio":[5000,23000]
               }
      print("seleccione el metodo de pago")
      print("1. efectivo")
      print("2. tarjeta de crédito")
      print("3.tarjeta débito")
      print("4. otro medio de pago")

      if total > 0:
        print("\n------- PAGO -------")
        print("Total a pagar: $", total)

        pago = int(input("Ingrese el dinero con el que va a pagar\n"))

        if pago >= total:
          cambio = pago - total
          print("Pago realizado correctamente")
          print("Su cambio es: $", cambio)

        else:
          print("El dinero no es suficiente")

      else:
        print("No hay productos para pagar")

      pass

    case 4:
      print("\n-------FACTURA -------")

      if len(carrito) > 0:
        for producto in carrito:
          print("Flor:", producto[0])
          print("Cantidad:", producto[1])
          print("Precio:", producto[2])
          print("Subtotal:", producto[3])
          print("-----------------------------")

        print("TOTAL: $", total)

      else:
        print("No hay productos en la factura")

      pass

    case 5:
      print("Gracias por su compra")
      break

    case _:
      print("Opción no válida")
```
