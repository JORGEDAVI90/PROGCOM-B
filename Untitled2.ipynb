```python
lista = [4,18,2,35,12,7,40,3]

nueva_lista = []

for numero in lista:
    if numero >= 10:
        nueva_lista.append(numero)

print(nueva_lista)
```

```python
objetos = ["Oro", "piedra", "madera", "diamante", "pan", "herramienta"]

resultado = []

for objeto in objetos:
    if len(objeto) > 5:
        resultado.append(objeto)

print(resultado)
```

```python
objetos = ["Oro", "piedra", "madera", "diamante", "pan", "herramienta"]

for objeto in objetos:
    if len(objeto) % 2 == 0:
        print(objeto)
```

```python
numeros = []

for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        numeros.append(i)

print(numeros)
```

```python
cantidades = [3, 15, 0, 8, 21, 2]

resultado = ["disponible" if cantidad > 0 else "agotado" for cantidad in cantidades]

print(resultado)
```

```python
numeros = list(range(1, 21))

resultado = ["par" if numero % 2 == 0 else "impar" for numero in numeros]

print(resultado)
```

```python
precios = [5000, 12000, 30000, 8000, 4500]

resultado = [precio * 0.8 if precio >= 10000 else precio for precio in precios]

print(resultado)
```
