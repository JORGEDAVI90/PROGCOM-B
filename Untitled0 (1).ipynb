#Estrctura de Selección


##Estructura if/else
Sólo posee dos valores de verdad
(V) o (F)

Sintaxis
```
if (condición)
  Se ejecutan operaciones si condicion es verdadera
else:
  Se ejecutan operaciones si condicion es falsa
```  

##if/elif/else
Evalua tantas condiciones como elif hayan. Si no se cumple, pasa a la siguiente condición, hasta llegar al else

Sintaxis
```
if (condición 1):
  aquí se ejecuta el código si condicional es V

elif (condición 2):
  se ejecuta si condicion 1 es falsa pero condición 2 es V

else(ninguna de las anteriores)
```

##Match/case
La estructura match evalua una variable frente a varios bloques case. El comodín case__funciona como ninguna de las anteriores.

Sintaxis
```
match(condición):
  case v1: aquí se ejecuta si condición==v1
  case v2: aquí se ejecuta si condición==v2
  case 3v: aquí se ejecuta si condición==v3
  case__: ninguna de las anteriores
```

````python
helado=input("Quieres helado?")
````

````python
#if/else
if helado.lower()=="si" or helado.lower()=="sí": print("pte lo compro")
else: print("no te lo compro")
````

````python
#match/case
match helado.lower():
  case "si": print("te lo compro")
  case "sipis": print("te lo compro")
  case "sí": print("te lo compro")
  case "siza": print("te lo compro")
  case __:print("no te lo compro")
````
