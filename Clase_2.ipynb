#Estructura de Repetición

son estructuras empleadas para repetir una acción hasta que se logre una condición

estructura paea (for)
for(variable) in range(VF)veces

```python
#que imprima los múltiplos de 3, que se encuentren dep 1 al 30
for intento in range(31):
  if intento%3==0:
    print(intento)
```

```python
for prueba in range(3,31,3):
  print(prueba)
```

```python
for despegue in range(10,-1,-1):
  print(despegue)
```

```python
for table in range(1,11):
  print("4*",table,"=",4*table)
```
