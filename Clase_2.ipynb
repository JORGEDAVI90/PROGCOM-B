#Estructura de Repetición

estructura paea (for)
for(variable) in range(VF)veces

````python
#que imprima los múltiplos de 3, que se encuentren dep 1 al 30
for intento in range(31):
  if intento%3==0:
    print(intento)
````

````python
#Nivel 3 Ejercisio 14
codigos = ["P01", "P02", "P03", "P04", "P05", "P06", "P07", "P08", "P09", "P10"]
nombres = ["Arroz", "Aceite", "Azucar", "Leche", "Cafe", "Pan", "Huevos", "Queso", "Jabon", "Papel"]
precios = [3500, 8500, 3200, 4200, 12000, 2500, 9000, 15000, 2800, 6000]

i = 0
while i < 10:
    print(codigos[i], nombres[i], precios[i])
    i = i + 1

total = 0
items = 0
seguir = True

while seguir == True:
    codigo = input("codigo o fin: ")

    if codigo == "fin":
        seguir = False
    else:
        pos = -1
        i = 0
        while i < 10:
            if codigos[i] == codigo:
                pos = i
            i = i + 1

        if pos == -1:
            print("no existe")
        else:
            cantidad = int(input("cantidad: "))
            subtotal = precios[pos] * cantidad
            total = total + subtotal
            items = items + 1
            print(nombres[pos], cantidad, subtotal)

if items == 0:
    print("no hubo compra")
else:
    iva = total * 0.19
    totalfinal = total + iva
    print("total", total)
    print("iva", iva)
    print("total final", totalfinal)

    pago = int(input("pago: "))

    while pago < totalfinal:
        print("no alcanza")
        pago = int(input("pago: "))

    cambio = pago - totalfinal
    print("cambio", cambio)
````
